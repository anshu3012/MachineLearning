# The standard normal distribution and the z-table
Standardizing a column, reading z-table values with scipy, two worked problems, the empirical rule, and z-score outliers.

```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go
from scipy import stats
```

## 1. Standardizing the Titanic ages

```python
df = pd.read_csv("data/titanic_train.csv")
age = df["Age"].dropna()
print(len(age), "ages, mean", round(age.mean(), 2), "std", round(age.std(), 2))
# subtract the mean, divide by the standard deviation
z = (age - age.mean()) / age.std()
print("after: mean", z.mean(), "std", z.std())
# the shape is unchanged: same skewness before and after
print("skewness", round(age.skew(), 3), round(z.skew(), 3))
```

```python
fig = go.Figure()
fig.add_histogram(x=z, nbinsx=30, histnorm="probability density", name="standardized ages")
grid = np.linspace(-3, 4, 300)
fig.add_scatter(x=grid, y=stats.norm.pdf(grid), name="N(0, 1)")
fig.update_layout(template="simple_white", xaxis_title="z")
fig.show()
```

## 2. The z-table with scipy
`cdf` is the area to the left of z, the number printed in a z-table.

```python
for value in [0, 1, 1.33, 2, 3, -1.33]:
    print(value, round(stats.norm.cdf(value), 5))
# standard normal PDF at 0 and 1
print(stats.norm.pdf(0), stats.norm.pdf(1))
```

## 3. Taller than 72 inches, heights N(68, 3^2)

```python
z72 = (72 - 68) / 3
print("z =", round(z72, 4))
# with z rounded to 1.33 as in a printed table
print("table:", round(1 - stats.norm.cdf(1.33), 5))
# exact
print("exact:", round(stats.norm(68, 3).sf(72), 5))
```

## 4. The empirical rule from the CDF

```python
for k in [1, 2, 3]:
    one_side = stats.norm.cdf(k) - 0.5
    print(f"within {k} sd: one side {one_side:.4f}, both sides {2 * one_side:.2%}")
```

## 5. Z-score outliers in the ages

```python
lower = age.mean() - 3 * age.std()
upper = age.mean() + 3 * age.std()
print("limits:", round(lower, 2), round(upper, 2))
print(age[(age < lower) | (age > upper)].sort_values().tolist())
```
