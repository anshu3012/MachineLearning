# Univariate Analysis

Analysis performed on a single variable or feature or col. Classify each col if it is a numerical or categorical col. Based on the classification we will use different Analysis

```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('train.csv')
```

```python
df.head()
```

# 1. Categorical Data

## a. Countplot - frequency count

```python
sns.countplot(x='Survived', data=df,palette='Set1')
df['Survived'].value_counts()
# df['Survived'].value_counts().plot(kind='bar') # to plot
```

```python
sns.countplot(x='Embarked', data=df)
```

```python
sns.countplot(x='Pclass', data=df)
```

```python
sns.countplot(x='Sex', data=df)
```

## b. PieChart

```python
df['Sex'].value_counts().plot(kind='pie',autopct='%.2f') # autopct shows percentage 
```

```python
df['Survived'].value_counts().plot(kind='pie',autopct='%.2f')
```
