# Maximum likelihood and binary cross entropy

```
import numpy as np
from sklearn.metrics import log_loss
```

## Two models, four points: probability each model gives to the true class

```
model_1 = np.array([0.7, 0.4, 0.4, 0.8])
model_2 = np.array([0.7, 0.6, 0.7, 0.6])
for name, p in [("model 1", model_1), ("model 2", model_2)]:
    print(f"{name}: likelihood {p.prod():.4f}, log-likelihood {np.log(p).sum():.3f}, cross entropy {-np.log(p).sum():.3f}")
```

## Why logs: products of many probabilities underflow

```
p = np.full(10_000, 0.7)
print("product:", p.prod())               # too small for a float: 0.0
print("sum of logs:", np.log(p).sum())
```

## The formula with y and y_hat, checked against scikit-learn

```
y = np.array([1, 0, 1, 0])                        # green = 1, red = 0
y_hat = np.array([0.7, 0.6, 0.4, 0.2])            # model 1's P(green) for each point

def bce(y, y_hat):
    return -np.mean(y * np.log(y_hat) + (1 - y) * np.log(1 - y_hat))

print("by hand:", round(bce(y, y_hat), 4), " sklearn log_loss:", round(log_loss(y, y_hat), 4))
```
