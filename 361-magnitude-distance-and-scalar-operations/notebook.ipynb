# Magnitude, distance and scalar operations

Code for the Note `note.md`: the length of a vector, the Euclidean distance, a nearest-neighbour demo, shifting and scaling, and mean centring.

```python
import numpy as np
import plotly.graph_objects as go
```

## 1. Magnitude: distance from the origin

```python
print(np.linalg.norm([3, 4]))        # 2D: 5.0
print(np.linalg.norm([2, 3, 6]))     # 3D: 7.0

# any number of dimensions works the same way
v = np.array([1, 2, 3, 4, 5])
print(np.linalg.norm(v))             # 7.416...
print(np.sqrt(np.sum(v ** 2)))       # the formula by hand, same answer
```

```python
# a 15-dimensional vector: no change to the code
rng = np.random.default_rng(1)
big = rng.integers(-5, 6, size=15)
print(big, np.linalg.norm(big))
```

## 2. Euclidean distance: subtract, then take the norm

```python
a = np.array([6, 7, 8, 9, 10])
b = np.array([1, 2, 3, 4, 5])
diff = a - b                  # component by component
print(diff)
print(np.linalg.norm(diff))   # 11.18...
```

## 3. Nearest neighbour in 3D
Five labelled vectors and a query. The query takes the class of the closest vector (the idea behind KNN).

```python
X = np.array([[3, 1, 3], [5, 4, 5], [6, 5, 4], [1, 2, 1], [5, 6, 6]])
y = np.array([0, 1, 1, 0, 1])

# change the query here to try other points
query = np.array([1, 1, 1])

# distance from the query to every vector at once
d = np.linalg.norm(X - query, axis=1)
print(d.round(2))
nearest = d.argmin()
print('nearest vector:', X[nearest], 'class:', y[nearest])
```

```python
fig = go.Figure()
for k, col in [(0, '#4C78A8'), (1, '#F58518')]:
    m = y == k
    fig.add_trace(go.Scatter3d(x=X[m, 0], y=X[m, 1], z=X[m, 2], mode='markers', name=f'class {k}',
                               marker=dict(size=7, color=col)))
fig.add_trace(go.Scatter3d(x=[query[0]], y=[query[1]], z=[query[2]], mode='markers', name='query',
                           marker=dict(size=8, color='#54A24B', symbol='diamond')))
fig.show()
```

## 4. Scalar operations: shifting and scaling

```python
print(np.array([2, 2]) + 5)    # shifting: [7 7]

v = np.array([2, 3])
print(v + 3)    # shifting: [5 6]
print(v - 1)    # shifting down: [1 2]
print(v * 2)    # scaling: [4 6]
print(v / 2)    # scaling down: [1.  1.5]

# scaling multiplies the length by the same factor
print(np.linalg.norm(v * 2) / np.linalg.norm(v))
```

## 5. Mean centring
First the small table from the Note, then 100 random 2D vectors.

```python
data = np.array([[3, 4], [5, 6], [7, 8]])
means = data.mean(axis=0)
print(means)               # [5. 6.]
centred = data - means     # one scalar subtracted per column
print(centred)
print(centred.sum(axis=0)) # each column now adds up to 0
```

```python
rng = np.random.default_rng(0)
pts = rng.multivariate_normal([6, 4], [[2, 1.2], [1.2, 1.5]], size=100)
cen = pts - pts.mean(axis=0)
print(pts.mean(axis=0).round(2), cen.mean(axis=0).round(10))

fig = go.Figure()
fig.add_trace(go.Scatter(x=pts[:, 0], y=pts[:, 1], mode='markers', name='original'))
fig.add_trace(go.Scatter(x=cen[:, 0], y=cen[:, 1], mode='markers', name='mean-centred'))
fig.update_xaxes(zeroline=True, zerolinecolor='black')
fig.update_yaxes(zeroline=True, zerolinecolor='black', scaleanchor='x')
fig.show()
```
