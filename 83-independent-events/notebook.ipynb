# Independent events

```
from fractions import Fraction
from itertools import product

outcomes = list(product(range(1, 7), repeat=2))
def P(event):
    return Fraction(sum(event(o) for o in outcomes), len(outcomes))

A = lambda o: o[1] == 6        # die 2 shows 6
B = lambda o: o[0] == 3        # die 1 shows 3
pA, pB, pAB = P(A), P(B), P(lambda o: A(o) and B(o))
print("P(A) =", pA, " P(B) =", pB, " P(A and B) =", pAB, " P(A) * P(B) =", pA * pB)
print("P(A | B) =", pAB / pB, "= P(A):", pAB / pB == pA)
```

## A pair that is not independent

```
C = lambda o: o[0] + o[1] >= 10   # sum at least 10
D = lambda o: o[0] == 6            # die 1 shows 6
pC, pD, pCD = P(C), P(D), P(lambda o: C(o) and D(o))
print("P(C) =", pC, " P(D) =", pD, " P(C and D) =", pCD, " P(C) * P(D) =", pC * pD)
print("P(C | D) =", pCD / pD)
```

## Coins have no memory

```
import numpy as np
rng = np.random.default_rng(0)
flips = rng.integers(0, 2, size=(1_000_000, 4))           # 1 = heads
three_heads = flips[:, :3].sum(axis=1) == 3
print("P(heads on flip 4 | heads on flips 1-3) ~", round(flips[three_heads, 3].mean(), 4), "from", three_heads.sum(), "cases")
```
