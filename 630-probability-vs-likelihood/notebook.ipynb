# Probability vs likelihood
Every number in the Note: the coin, the bag, the heights, the mouse weights, and why a likelihood does not add up to 1.

```python
import numpy as np
import plotly.graph_objects as go
from scipy import stats
from scipy.integrate import quad
```

## 1. Coin: probability of tails, likelihood of p after five heads

```python
p = 0.5
# Bernoulli PMF p^k (1 - p)^(1 - k); tails is k = 0
print("P(tails | p = 0.5) =", p ** 0 * (1 - p) ** 1)
# likelihood of p given five heads: p^5
for p in (0.5, 0.7):
    print(f"L(p = {p} | 5 heads) = {p ** 5:.3f}")
```

## 2. Bag with 3 red and 2 green balls

```python
p_green = 2 / 5
print("P(red) =", 1 - p_green)
for p in (2 / 5, 4 / 5):
    print(f"L(p = {p:.1f} | 5 green) = {p ** 5:.3f}")
```

## 3. Heights: probability is an area, likelihood is a height

```python
heights = stats.norm(150, 10)
print("P(170 <= X <= 180) =", round(heights.cdf(180) - heights.cdf(170), 4))
for x in (100, 130, 140, 150, 200):
    print(f"L(mu = 150, sigma = 10 | x = {x}) = {heights.pdf(x):.3g}")
```

The likelihood function: one person at 130 cm, slide the mean.

```python
mus = np.linspace(100, 200, 401)
L = stats.norm(mus, 10).pdf(130)
print("likelihood is highest at mu =", mus[L.argmax()], "with value", L.max().round(4))
fig = go.Figure(go.Scatter(x=mus, y=L, mode="lines"))
fig.update_layout(template="simple_white", xaxis_title="mu (cm)", yaxis_title="likelihood given 130 cm", height=350)
fig.show()
```

## 4. Mouse weights (StatQuest)

```python
mice = stats.norm(32, 2.5)
print("P(32 <= weight <= 34) =", round(mice.cdf(34) - mice.cdf(32), 2))
print("L(mu = 32 | 34 g) =", round(mice.pdf(34), 2), "  L(mu = 34 | 34 g) =", round(stats.norm(34, 2.5).pdf(34), 2))
```

## 5. Probabilities add up to 1, likelihoods do not

```python
# probabilities of 0, 1, ..., 5 heads in 5 tosses with p = 0.5
print("sum of probabilities over all events:", stats.binom(5, 0.5).pmf(range(6)).sum())
# area under the likelihood p^5 over all p from 0 to 1
area, _ = quad(lambda p: p ** 5, 0, 1)
print("area under the likelihood p^5:", round(area, 3))
```
