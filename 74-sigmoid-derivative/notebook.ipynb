# The derivative of the sigmoid

```
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def sigmoid_derivative(z):
    return sigmoid(z) * (1 - sigmoid(z))
```

## Check against a numerical slope: (sigmoid(z + h) - sigmoid(z - h)) / 2h

```
h = 1e-5
for z in [-4, -2, 0, 2, 4]:
    numeric = (sigmoid(z + h) - sigmoid(z - h)) / (2 * h)
    print(f"z = {z:2d}: formula {sigmoid_derivative(z):.6f}, numerical {numeric:.6f}")
```

## The largest value is 0.25, at z = 0

```
z = np.linspace(-10, 10, 100001)
d = sigmoid_derivative(z)
print("max", d.max(), "at z =", z[d.argmax()])
```
