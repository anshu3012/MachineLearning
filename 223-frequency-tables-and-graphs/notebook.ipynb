# Frequency tables and graphs by data type
The tables and charts of the Note, built with pandas and Plotly.

```python
import pandas as pd
import plotly.express as px

# the 200-person vacation survey, rebuilt as one row per person
counts = {'Beach': 60, 'City': 40, 'Adventure': 30, 'Nature': 35, 'Cruise': 20, 'Other': 15}
survey = pd.DataFrame({'vacation': [k for k, n in counts.items() for _ in range(n)]})
survey.shape
```

## 1. Frequency, relative frequency, cumulative frequency

```python
# count each category, keeping the survey's order
freq = survey['vacation'].value_counts().reindex(list(counts))
table = pd.DataFrame({'frequency': freq,
                      'relative': freq / freq.sum(),
                      'cumulative': freq.cumsum()})
table['cumulative relative'] = table['relative'].cumsum()
table
```

```python
# frequency as a bar chart, relative frequency as a pie, cumulative as a line
px.bar(table, y='frequency').show()
px.pie(table, names=table.index, values='relative').show()
px.line(table, y='cumulative', markers=True).show()
```

## 2. Histogram: change the number of bins

```python
df = pd.read_csv('data/titanic_train.csv')
# try nbins = 2, 10, 80 and compare the shapes
px.histogram(df, x='Age', nbins=10)
```

## 3. Categorical + numerical: any aggregate per category

```python
df.groupby('Sex')['Age'].agg(['mean', 'median', 'max'])
```

```python
# the bar chart of one aggregate: the maximum age
px.bar(df.groupby('Sex', as_index=False)['Age'].max(), x='Sex', y='Age')
```

## 4. Categorical + numerical: contingency table of bins

```python
# cut the ages into bands, then count each band and sex
bands = pd.cut(df['Age'], bins=[0, 10, 20, 30, 40, 50, 60, 80])
pd.crosstab(bands, df['Sex'])
```

```python
# a pivot table aggregates a third column: mean age by class and sex
df.pivot_table(index='Pclass', columns='Sex', values='Age', aggfunc='mean').round(1)
```
