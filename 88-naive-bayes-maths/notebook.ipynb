# The Naive Bayes formula: exact chain rule versus the naive product

```
import numpy as np
rng = np.random.default_rng(0)
n = 200_000
# Two classes; within each class the two binary inputs are drawn INDEPENDENTLY
y = rng.random(n) < 0.4
x1 = np.where(y, rng.random(n) < 0.8, rng.random(n) < 0.3)
x2 = np.where(y, rng.random(n) < 0.6, rng.random(n) < 0.1)

def compare(x1, x2, label):
    c = y
    exact = ((x1 & x2)[c]).mean()                     # P(x1=1, x2=1 | C=1)
    naive = x1[c].mean() * x2[c].mean()               # P(x1=1 | C=1) P(x2=1 | C=1)
    chain = (x1[c & x2]).mean() * x2[c].mean()        # P(x1 | x2, C) P(x2 | C): the exact chain rule
    print(f"{label}: exact {exact:.4f}  chain rule {chain:.4f}  naive product {naive:.4f}")

compare(x1, x2, "independent within the class")
x2_copy = x1.copy()                                   # a second input that is just a copy of the first
compare(x1, x2_copy, "x2 is a copy of x1           ")
```

## The MAP rule on the eight matches

```
import pandas as pd
df = pd.read_csv("../87-naive-bayes-intuition/data/matches.csv")
query = {"toss": "lost", "venue": "Mumbai", "outlook": "sunny"}
scores = {}
for c in df["result"].unique():
    sub = df[df["result"] == c]
    scores[c] = len(sub) / len(df) * np.prod([(sub[col] == v).mean() for col, v in query.items()])
print(scores, "-> argmax:", max(scores, key=scores.get))
```
