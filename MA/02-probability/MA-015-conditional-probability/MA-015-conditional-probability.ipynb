# Conditional probability with two dice

```
from fractions import Fraction
from itertools import product

outcomes = list(product(range(1, 7), repeat=2))     # (die 1, die 2)
print(len(outcomes), "outcomes")

def P(event, given=None):
    space = [o for o in outcomes if given is None or given(o)]
    return Fraction(sum(event(o) for o in space), len(space))

A = lambda o: o[0] == 5              # die 1 shows 5
B = lambda o: o[0] + o[1] <= 10      # sum at most 10
print("P(A)       =", P(A))
print("P(B)       =", P(B))
print("P(A and B) =", P(lambda o: A(o) and B(o)))
print("P(A | B) counted in the reduced space =", P(A, given=B))
print("P(A | B) from the formula             =", P(lambda o: A(o) and B(o)) / P(B))
```

## A check by simulation

```
import numpy as np
rng = np.random.default_rng(0)
rolls = rng.integers(1, 7, size=(1_000_000, 2))
b = rolls.sum(axis=1) <= 10
print("simulated P(A | B):", round((rolls[b, 0] == 5).mean(), 4), " exact:", round(5 / 33, 4))
```
