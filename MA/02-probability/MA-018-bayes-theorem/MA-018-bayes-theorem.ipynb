# Bayes' theorem

```
from fractions import Fraction
from itertools import product

outcomes = list(product(range(1, 7), repeat=2))
def P(event):
    return Fraction(sum(event(o) for o in outcomes), len(outcomes))

A = lambda o: o[0] == 5                 # die 1 shows 5
B = lambda o: o[0] + o[1] <= 10         # sum at most 10
p_B_given_A = P(lambda o: A(o) and B(o)) / P(A)
print("P(B | A) =", p_B_given_A)
print("Bayes:  P(A | B) = P(B | A) P(A) / P(B) =", p_B_given_A * P(A) / P(B))
print("Direct: P(A | B) =", P(lambda o: A(o) and B(o)) / P(B))
```

## A first look at spam

```
p_spam = 0.20                 # prior: 20% of emails are spam
p_free_given_spam = 0.60      # likelihood: 'free' appears in 60% of spam
p_free_given_ham = 0.05       # ... and in 5% of normal emails
p_free = p_free_given_spam * p_spam + p_free_given_ham * (1 - p_spam)    # evidence
print("P(free) =", round(p_free, 3))
print("P(spam | free) =", round(p_free_given_spam * p_spam / p_free, 3))
```
