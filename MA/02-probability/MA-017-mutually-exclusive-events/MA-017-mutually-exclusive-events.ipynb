# Mutually exclusive events

```
from fractions import Fraction
from itertools import product

outcomes = list(product(range(1, 7), repeat=2))
def P(event):
    return Fraction(sum(event(o) for o in outcomes), len(outcomes))

A = lambda o: o[0] == 3        # die 1 shows 3
B = lambda o: o[0] == 6        # die 1 shows 6
pA, pB, pAB = P(A), P(B), P(lambda o: A(o) and B(o))
print("P(A) =", pA, " P(B) =", pB, " P(A and B) =", pAB)
print("P(A | B) =", pAB / pB, "   P(A) * P(B) =", pA * pB, "-> not independent")
print("P(A or B) =", P(lambda o: A(o) or B(o)), "= P(A) + P(B) =", pA + pB)
```
