# A Bayes' theorem problem: which machine made the defective marker?

```
import numpy as np

prior = {"M1": 0.20, "M2": 0.30, "M3": 0.50}          # share of production
p_defect = {"M1": 0.05, "M2": 0.03, "M3": 0.01}       # P(defective | machine)

joint = {m: prior[m] * p_defect[m] for m in prior}    # P(defective and machine)
p_D = sum(joint.values())                             # total probability
print("P(D) =", round(p_D, 4))
for m in prior:
    print(f"P({m} | D) = {joint[m]:.3f} / {p_D:.3f} = {joint[m] / p_D:.3f}")
```

## Check by simulating a million markers

```
rng = np.random.default_rng(0)
n = 1_000_000
machine = rng.choice(3, size=n, p=[0.2, 0.3, 0.5])
defective = rng.random(n) < np.array([0.05, 0.03, 0.01])[machine]
print("simulated P(D)      ", round(defective.mean(), 4))
print("simulated P(M3 | D) ", round((machine[defective] == 2).mean(), 4))
```
