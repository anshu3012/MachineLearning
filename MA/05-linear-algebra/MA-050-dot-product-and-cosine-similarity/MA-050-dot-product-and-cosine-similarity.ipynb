# Dot product and cosine similarity

Code for the Note `note.md`: the dot product three ways, its two laws, the angle between vectors, and cosine similarity for a tiny recommender.

```python
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

## 1. The dot product: multiply matching components, then add

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(np.sum(a * b))   # by hand: 4 + 10 + 18
print(np.dot(a, b))    # NumPy's function
print(a @ b)           # the @ operator
```

```python
# as a matrix product: row (1 x 3) times column (3 x 1)
row = a.reshape(1, -1)
col = b.reshape(-1, 1)
print(row @ col, (row @ col).shape)   # [[32]], a 1 x 1 matrix
```

## 2. Commutative and distributive laws

```python
c = np.array([7, 8, 9])
print(np.dot(a, b), np.dot(b, a))                 # commutative: both 32
print(np.dot(a, b + c), np.dot(a, b) + np.dot(a, c))  # distributive: both 82
```

## 3. The geometric formula and the angle

```python
u = np.array([3, 4])
v = np.array([4, 3])
dot = u @ v
norms = np.linalg.norm(u) * np.linalg.norm(v)
cos = dot / norms
print(dot, norms, cos)
print('angle in degrees:', np.degrees(np.arccos(cos)))

# perpendicular vectors: dot product 0
print(np.array([3, 4]) @ np.array([-4, 3]))
```

```python
# check a . b = |a||b|cos(theta) for the animation's 30-degree case
A = np.array([3.0, 1.0])
t = np.arctan2(1, 3) + np.radians(30)
B = 2.5 * np.array([np.cos(t), np.sin(t)])
print(A @ B, np.linalg.norm(A) * 2.5 * np.cos(np.radians(30)))
```

## 4. Cosine similarity

```python
def cos_sim(x, y):
    """Cosine of the angle between x and y."""
    return x @ y / (np.linalg.norm(x) * np.linalg.norm(y))

p = np.array([1, 2, 3])
q = np.array([2, 4, 5])
r = np.array([-1, -2, 1])
print(round(cos_sim(p, q), 3), round(np.degrees(np.arccos(cos_sim(p, q))), 1))
print(round(cos_sim(p, r), 3), round(np.degrees(np.arccos(cos_sim(p, r))), 1))
print(cos_sim(p, 10 * p))   # same direction, any length: 1
```

## 5. A recommender with cosine similarity
The three toy summaries from the vectors Note, plus a longer copy of B to show that length does not matter.

```python
texts = ['hi how are you', 'my name is riya', 'this is 2023', 'my name is riya my name is riya']
X = CountVectorizer().fit_transform(texts).toarray()

# cosine similarity of every text with every other
S = cosine_similarity(X)
print(S.round(3))

# Euclidean distance from B for comparison
print(np.linalg.norm(X - X[1], axis=1).round(2))
```
