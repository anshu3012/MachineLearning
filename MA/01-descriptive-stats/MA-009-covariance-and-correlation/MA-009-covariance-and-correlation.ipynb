# Covariance and correlation
The worked examples of the Note, and the scale experiment.

```python
import numpy as np
import pandas as pd
import plotly.express as px
```

## 1. Covariance by hand

```python
x = np.array([2, 5, 8, 12, 13])      # experience, years
y = np.array([1, 2, 5, 12, 10])      # salary, lakh rupees a month
# distances from the two means, and their products
table = pd.DataFrame({'x': x, 'y': y, 'dx': x - x.mean(), 'dy': y - y.mean()})
table['product'] = table.dx * table.dy
print(table)
# sample covariance: divide the sum by n - 1
print('covariance', table['product'].sum() / (len(x) - 1), np.cov(x, y)[0, 1])
```

```python
# the backlogs data: a negative relationship
backlogs = np.array([2, 5, 8, 12, 13])
package = np.array([10, 12, 5, 2, 1])
print(np.cov(backlogs, package)[0, 1], np.corrcoef(backlogs, package)[0, 1])
# everyone gets the same package: covariance exactly 0
print(np.cov(backlogs, np.full(5, 10))[0, 1])
```

## 2. Correlation from covariance

```python
r = np.cov(x, y)[0, 1] / (x.std(ddof=1) * y.std(ddof=1))
print(r, np.corrcoef(x, y)[0, 1])
```

## 3. Changing the units
Covariance changes with the units; correlation does not.

```python
for name, a, b in [('years, lakhs', x, y), ('months, lakhs', 12 * x, y), ('years, rupees', x, 100_000 * y)]:
    print(f'{name:14s} cov {np.cov(a, b)[0, 1]:>12.1f}   r {np.corrcoef(a, b)[0, 1]:.3f}')
```

## 4. The scale experiment of Figure 2
Change the noise (45) or the factor (2) and rerun.

```python
rng = np.random.default_rng(42)       # fixed seed: same data every run
xs = rng.uniform(0, 120, 40)
ys = xs + rng.normal(0, 45, 40)
for name, a, b in [('x vs x', xs, xs), ('x vs y', xs, ys), ('2x vs 2y', 2 * xs, 2 * ys)]:
    print(f'{name:9s} cov {np.cov(a, b)[0, 1]:7.0f}   r {np.corrcoef(a, b)[0, 1]:.2f}')
# the covariance of x with itself is its variance
print('variance of x', xs.var(ddof=1))
```

```python
px.scatter(x=xs, y=ys, trendline='ols')
```
