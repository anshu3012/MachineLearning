# Type I and Type II errors, power, and one- vs two-tailed tests
Power of the training-program z-test, the alpha-beta trade-off, and two simulations.

```python
import numpy as np
from scipy import stats
```

## 1. Power if the true mean is 52

```python
mu0, mu_true, sigma = 50, 52, 5

def power_right(n, alpha):
    # shift of the z statistic, in standard errors, when the true mean is mu_true
    d = (mu_true - mu0) / (sigma / np.sqrt(n))
    beta = stats.norm.cdf(stats.norm.ppf(1 - alpha) - d)
    return 1 - beta

for alpha in [0.05, 0.01]:
    p = power_right(30, alpha)
    print(f"alpha {alpha}: beta {1 - p:.2f}, power {p:.2f}")
```

## 2. More data raises the power

```python
for n in [30, 60, 100]:
    print(f"n = {n:3d}: power {power_right(n, 0.05):.2f}")
```

## 3. Two-tailed power for the same effect

```python
d = (mu_true - mu0) / (sigma / np.sqrt(30))
c = stats.norm.ppf(0.975)
# reject in either tail
print(f"two-tailed power {stats.norm.sf(c - d) + stats.norm.cdf(-c - d):.2f}")
```

## 4. Simulation: the Type I error rate is alpha
We draw many samples from a population where H0 is true and count the rejections.

```python
rng = np.random.default_rng(0)
# 100,000 samples of 30 values each, true mean 50, so H0 is true
samples = rng.normal(50, 5, size=(100_000, 30))
z = (samples.mean(axis=1) - 50) / (5 / np.sqrt(30))
print("right-tailed test rejects:", np.mean(z > 1.645))
print("two-tailed test rejects:  ", np.mean(np.abs(z) > 1.96))
# choosing the tail after looking at z = rejecting when |z| > 1.645
print("tail chosen after looking:", np.mean(np.abs(z) > 1.645))
```

## 5. Simulation: power when the true mean is 52

```python
samples = rng.normal(52, 5, size=(100_000, 30))
z = (samples.mean(axis=1) - 50) / (5 / np.sqrt(30))
print("right-tailed power:", np.mean(z > 1.645))
```
