# Backpropagation: why the update works
The loss of the 2-2-1 regression network is a function of its 9 parameters. We freeze 8 of them and look at the
loss as a function of the output bias b21 alone, then run the update rule on it with four learning rates.

```python
import numpy as np
import pandas as pd

x, y = np.array([8.0, 8.0]), 4.0          # student 1: CGPA 8, profile score 8, package 4

def y_hat(W111, W121, W112, W122, b11, b12, W211, W221, b21, x=x):
    # the network written out in full: y_hat as a function of all 9 parameters
    O11 = W111 * x[0] + W121 * x[1] + b11
    O12 = W112 * x[0] + W122 * x[1] + b12
    return W211 * O11 + W221 * O12 + b21

def loss(*params):
    return (y - y_hat(*params)) ** 2

start = dict(W111=0.1, W121=0.1, W112=0.1, W122=0.1, b11=0, b12=0, W211=0.1, W221=0.1, b21=0)
print("loss at the start:", loss(*start.values()))
```

## Freeze 8 parameters: the loss as a function of b21 only
L(b21) = (4 - 0.32 - b21)^2, a parabola with its lowest point at b21 = 3.68.

```python
def L_b(b):
    return loss(*{**start, "b21": b}.values())

def slope_b(b):
    return -2 * (y - y_hat(*{**start, "b21": b}.values()))

for b in [-5, 0, 3.68, 5]:
    print(f"b21 = {b:5.2f}: loss {L_b(b):7.3f}, slope {slope_b(b):7.3f}")
```

## The update rule b21 <- b21 - lr * slope, from b21 = -5

```python
rows = []
for lr in [1.0, 0.1, 0.01, 1.1]:
    b = -5.0
    for step in range(11):
        rows.append(dict(lr=lr, step=step, b21=b, loss=L_b(b)))
        b = b - lr * slope_b(b)
paths = pd.DataFrame(rows)
paths.to_csv("data/b21_paths.csv", index=False)
for lr, g in paths.groupby("lr"):
    print(f"lr {lr}: b21 =", g.b21.round(3).tolist()[:6], "... step 10:", round(g.b21.iloc[-1], 3))
```
