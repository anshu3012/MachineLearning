# Ordinal encoding and label encoding
Turning the ordinal input columns `review` and `education` into numbers with `OrdinalEncoder`, and the target `purchased` with `LabelEncoder`.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OrdinalEncoder, LabelEncoder
```

## Load the data
50 customers: age, gender, their review of a product, their education, and whether they bought a recommended product.

```python
df = pd.read_csv("data/customer.csv")
df.sample(5, random_state=1)
```

```python
# the categories in each text column
for col in ["gender", "review", "education", "purchased"]:
    print(col, sorted(df[col].unique()))
```

## Keep review, education and purchased
`gender` is nominal and needs one-hot encoding (next Note); `age` is already a number. We drop both for now.

```python
# keep the columns from position 2 onwards: review, education, purchased
df = df.iloc[:, 2:]
df.head()
```

## Split first, then encode

```python
X_train, X_test, y_train, y_test = train_test_split(
    df.iloc[:, 0:2],   # inputs: review, education
    df.iloc[:, -1],    # target: purchased
    test_size=0.2, random_state=0)
X_train.shape, X_test.shape
```

```python
X_train.head()
```

## OrdinalEncoder on the inputs
One list per column, each written from lowest to highest.

```python
oe = OrdinalEncoder(categories=[
    ["Poor", "Average", "Good"],   # review
    ["School", "UG", "PG"],        # education
])
# fit: learn the categories from the training set
oe.fit(X_train)
# transform: replace each category by its position in the list
X_train_enc = oe.transform(X_train)
X_test_enc = oe.transform(X_test)
oe.categories_
```

```python
# original and encoded side by side
pd.concat([X_train.reset_index(drop=True),
           pd.DataFrame(X_train_enc, columns=["review_enc", "education_enc"])],
          axis=1).head()
```

## What happens without `categories`
The encoder sorts the categories alphabetically, which breaks the real order.

```python
oe_default = OrdinalEncoder().fit(X_train)
oe_default.categories_
```

## LabelEncoder on the target

```python
le = LabelEncoder()
# fit: learn the classes (sorted alphabetically)
le.fit(y_train)
le.classes_
```

```python
y_train_enc = le.transform(y_train)
y_test_enc = le.transform(y_test)
y_train.head().tolist(), y_train_enc[:5]
```

```python
# going back from numbers to labels
le.inverse_transform([0, 1])
```
