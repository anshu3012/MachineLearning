# Batch learning: how often should we retrain?

Run the cell with **Shift + Enter**, then move the slider.

A batch model is frozen after training, while the world keeps changing, so its performance slowly drops.
Retraining brings it back up. **Try:** compare retraining every day, every week and every month.
More frequent retraining keeps the model fresh, but every retrain costs time and computing power.

```python
import numpy as np
import plotly.graph_objects as go

days = np.linspace(0, 60, 3000)
choices = [1, 3, 7, 14, 30]                       # retrain every N days

def performance(retrain_every, decay=0.006, start=0.92):
    since_retrain = np.mod(days, retrain_every)    # days since the last retrain
    return start - decay * since_retrain           # performance drops a little each day

fig = go.Figure()
for i, n in enumerate(choices):
    fig.add_trace(go.Scatter(x=days, y=performance(n), mode="lines", visible=(n == 7),
                             line=dict(width=3), name=f"every {n} days"))

steps = [dict(method="update", label=str(n), args=[{"visible": [j == i for j in range(len(choices))]}])
         for i, n in enumerate(choices)]
fig.update_layout(sliders=[dict(active=2, steps=steps, currentvalue=dict(prefix="Retrain every (days): "))],
                  xaxis_title="Days since deployment", yaxis_title="Performance",
                  yaxis_range=[0.7, 0.95], height=500, template="simple_white", showlegend=False)
fig.show()
```
