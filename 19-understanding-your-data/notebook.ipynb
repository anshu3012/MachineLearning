# Understanding your data: seven first questions

Run each cell in order with **Shift + Enter**. Each step matches a section of the Note.
The data is the Titanic training set (`data/titanic_train.csv`): 891 passengers, 12 columns.

## 1. Load the data

```python
import pandas as pd

pd.__version__   # this Notebook was written for pandas 3.x
```

```python
# read the Titanic passengers into a DataFrame
df = pd.read_csv("data/titanic_train.csv")
```

## 2. How big is the data?

```python
# (rows, columns)
df.shape
```

## 3. What does the data look like?

```python
# the first 5 rows: can mislead if the file is sorted
df.head()
```

```python
# 5 random rows: a fairer first look
# random_state fixes the draw, so we get the same rows every run
df.sample(5, random_state=1)
```

```python
# what head() would show if the file were sorted by age: only small children
df.sort_values("Age").head()[["Name", "Age"]]
```

## 4. What type is each column?

```python
# column names, non-missing counts, dtypes and memory use
df.info()
```

```python
# these four columns hold small whole numbers but use 8 bytes per value (int64)
small_cols = ["Survived", "Pclass", "SibSp", "Parch"]
before = df[small_cols].memory_usage(index=False).sum()

# int8 holds -128 to 127 in 1 byte, plenty for these columns
after = df[small_cols].astype("int8").memory_usage(index=False).sum()
before, after
```

```python
# Age is float64: it has missing values (NaN is a float), and some ages have fractions
age = df["Age"].dropna()
age[age % 1 != 0].head(8).tolist()
```

## 5. Are there missing values?

```python
# number of missing values in each column
df.isnull().sum()
```

```python
# the same as a share of all rows (0.77 = 77%)
df.isnull().mean().round(3)
```

## 6. What does the data look like in numbers?

```python
# count, mean, std, min, 25%, 50%, 75%, max of every numeric column
df.describe()
```

```python
# the same summary for the text columns: count, unique, top, freq
df.describe(include="str")
```

```python
# a check on the worked example: five ages
ages = pd.Series([22, 38, 26, 35, 35])
ages.mean(), ages.std(), ages.quantile([0.25, 0.5, 0.75]).tolist()
```

```python
# an odd value describe() points us to: tickets with fare 0
(df["Fare"] == 0).sum()
```

## 7. Are there duplicate rows?

```python
# number of rows that are exact copies of an earlier row
df.duplicated().sum()
```

```python
# a copy with the first two rows added again, to see duplicates being found and removed
doubled = pd.concat([df, df.head(2)])
print(doubled.duplicated().sum())        # 2
print(doubled.drop_duplicates().shape)   # back to (891, 12)
```

## 8. How are the columns related?

```python
# correlation of every numeric column with Survived
# numeric_only=True skips text columns (pandas 2+ raises an error without it)
df.corr(numeric_only=True)["Survived"].round(2)
```

```python
# the full table: every numeric column against every other
df.corr(numeric_only=True).round(2)
```

```python
# Sex is text, so corr() skipped it; turned into 1 (female) and 0 (male) it is the strongest link
df["Sex"].map({"female": 1, "male": 0}).corr(df["Survived"]).round(2)
```

```python
# the worked example: three passengers, class 1, 2, 3, fares 80, 20, 10
pd.Series([1, 2, 3]).corr(pd.Series([80, 20, 10])).round(3)
```
