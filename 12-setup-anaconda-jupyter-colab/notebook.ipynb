## 2. Library versions

`importlib.metadata.version` reads the installed version of a package by its install name,
without importing it. A package that is missing prints `not installed`.

```python
from importlib.metadata import version, PackageNotFoundError

# Install names of the key libraries, and the version pinned in environment.yml
pinned = {
    "numpy": "2.5.3",
    "pandas": "3.0.6",
    "scikit-learn": "1.9.1",
    "scipy": "1.18.1",
    "statsmodels": "0.15.0",
    "seaborn": "0.13.2",
    "plotly": "7.1.0",
    "dash": "4.4.1",
    "manim": "0.20.1",
    "jupyterlab": "4.6.4",
    "fg-data-profiling": "4.20.0",
}

for name, pin in pinned.items():
    try:
        found = version(name)
    except PackageNotFoundError:
        found = "not installed"
    # Mark every library whose version differs from the pin
    mark = "" if found == pin else "   <- differs"
    print(f"{name:18} {found:14} (pinned {pin}){mark}")
```

On Colab or Kaggle, many lines say `differs` or `not installed`: those services come with their own versions.
Locally, every line should match.

## 3. The pandas setting for the profiling report

pandas 3 stores text columns as Arrow arrays. The profiling library cannot add those up,
so before building a report we switch text storage back to plain Python strings.

```python
import pandas as pd

pd.set_option("mode.string_storage", "python")

# A tiny table with one text column, to see the setting at work
df = pd.DataFrame({"city": ["Delhi", "Pune"], "rain_mm": [12, 30]})
print(pd.get_option("mode.string_storage"))
print(df.dtypes)
```
