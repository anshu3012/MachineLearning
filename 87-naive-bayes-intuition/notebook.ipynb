# Naive Bayes by hand on eight matches

```
import pandas as pd
df = pd.read_csv("data/matches.csv")
df
```

## The exact likelihood is often zero

```
query = {"toss": "lost", "venue": "Mumbai", "outlook": "sunny"}
for cls in ["win", "loss"]:
    sub = df[df["result"] == cls]
    exact = ((sub["toss"] == "lost") & (sub["venue"] == "Mumbai") & (sub["outlook"] == "sunny")).mean()
    print(f"P(lost, Mumbai, sunny | {cls}) = {exact:.3f}   ({len(sub)} matches)")
```

## The naive version: multiply one-column probabilities

```
scores = {}
for cls in ["win", "loss"]:
    sub = df[df["result"] == cls]
    prior = len(sub) / len(df)
    parts = [(sub[col] == val).mean() for col, val in query.items()]
    scores[cls] = prior * parts[0] * parts[1] * parts[2]
    print(f"{cls}: prior {prior:.3f} x {' x '.join(f'{p:.3f}' for p in parts)} = {scores[cls]:.4f}")
total = sum(scores.values())
print({c: round(s / total, 3) for c, s in scores.items()}, "-> predict", max(scores, key=scores.get))
```
