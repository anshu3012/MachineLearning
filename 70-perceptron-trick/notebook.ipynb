# The perceptron trick: one update by hand

```
import numpy as np

w = np.array([5.0, 2.0, 3.0])        # line 5 + 2x + 3y = 0, as [w0, w1, w2]

def side(w, point):
    value = w @ np.r_[1, point]      # put 1 in front for the intercept
    return value, "positive" if value > 0 else ("negative" if value < 0 else "on the line")

for p in [(2, 1), (-4, -3), (-1, -1)]:
    print(p, side(w, p))
```

## A negative point on the positive side: subtract

```
p = np.array([5, 2])
print("before", side(w, p))
w_new = w - np.r_[1, p]
print("full step:", w_new, side(w_new, p))
w_small = w - 0.1 * np.r_[1, p]
print("learning rate 0.1:", w_small, side(w_small, p))
```

## A positive point on the negative side: add

```
p = np.array([-3, -2])
print("before", side(w, p))
w_new = w + np.r_[1, p]
print("full step:", w_new, side(w_new, p))
```

## One rule for all four cases: w = w + lr * (y - y_hat) * x

```
def step(z):
    return 1 if z > 0 else 0

lr = 1.0
for y_true, p in [(0, (5, 2)), (1, (-3, -2)), (1, (2, 1)), (0, (-4, -3))]:
    x = np.r_[1, p]
    y_hat = step(w @ x)
    print(f"y={y_true}, y_hat={y_hat}, y - y_hat = {y_true - y_hat:+d} -> new w = {w + lr * (y_true - y_hat) * x}")
```
